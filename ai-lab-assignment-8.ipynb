# Write a Python Program to Implement the Minimax algorithm for a two-player game.

1. Implement the Minimax algorithm using recursion.
2. Consider MAX and MIN players.
3. Use the following leaf-node values:

   3, 5, 2, 9, 12, 5, 23, 23

4. Determine the optimal score for the MAX player.
5. Display the final optimal score as output.

```python

import math

def minimax(depth, node_index, is_maximizing, values, max_depth):

    if depth == max_depth:
        return values[node_index]

    if is_maximizing:
        return max(
            minimax(depth + 1, node_index * 2, False, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, False, values, max_depth)
        )

    else:
        return min(
            minimax(depth + 1, node_index * 2, True, values, max_depth),
            minimax(depth + 1, node_index * 2 + 1, True, values, max_depth)
        )

leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]

tree_depth = int(math.log2(len(leaf_values)))

best_score = minimax(0, 0, True, leaf_values, tree_depth)

print("Best possible score:", best_score)
```

# write a python program to implement the Alpha-Beta Pruning algorithm for a two-player game.Use the leaf-node values3,5,2,9,12,5,23,23. Determine the optimal score for the MAX player and display which branches are pruned during the search.

```python
import math

def alpha_beta(depth, node_index, maximizing_player, values, alpha, beta, max_depth):

    # Base case: leaf node
    if depth == max_depth:
        return values[node_index]

    if maximizing_player:
        # MAX player
        best = float('-inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                values,
                alpha,
                beta,
                max_depth
            )

            best = max(best, value)
            alpha = max(alpha, best)

            # Alpha-Beta pruning
            if beta <= alpha:
                print("Branch pruned at node", node_index)
                break

        return best

    else:
        # MIN player
        best = float('inf')

        for i in range(2):
            value = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                values,
                alpha,
                beta,
                max_depth
            )

            best = min(best, value)
            beta = min(beta, best)

            # Alpha-Beta pruning
            if beta <= alpha:
                print("Branch pruned at node", node_index)
                break

        return best


# Leaf node values
leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]

# Calculate maximum depth
max_depth = int(math.log2(len(leaf_values)))

# Initial alpha and beta values
alpha = float('-inf')
beta = float('inf')

# Find optimal score for MAX
best_score = alpha_beta(
    0,
    0,
    True,
    leaf_values,
    alpha,
    beta,
    max_depth
)

print("Optimal score for MAX:", best_score)
```
